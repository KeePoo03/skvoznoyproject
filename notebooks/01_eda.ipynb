```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/student-por.csv", sep=";")
print("Размер:", df.shape)
print("Дубликатов:", df.duplicated().sum())
print("Пропусков всего:", df.isna().sum().sum())
display(df.head())
print(df.dtypes)
```

```python
print(df[["age", "studytime", "failures", "absences", "G1", "G2", "G3"]].describe().round(2))
print("Асимметрия G3 (skew):", round(df["G3"].skew(), 3))

fig, ax = plt.subplots(1, 2, figsize=(10, 3.5))
df["G3"].hist(bins=21, ax=ax[0])
ax[0].set_title("Распределение итоговой оценки G3")
ax[0].set_xlabel("G3")

ax[1].scatter(df["G2"], df["G3"], alpha=0.4)
ax[1].set_title("Связь G2 и G3")
ax[1].set_xlabel("G2")
ax[1].set_ylabel("G3")
plt.tight_layout()
plt.show()
```
